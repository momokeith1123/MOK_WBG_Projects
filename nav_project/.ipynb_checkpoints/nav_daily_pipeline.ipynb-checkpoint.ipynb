```python
# ==== CELL 1: Imports and configuration ====
import re
import json
from functools import reduce
from pathlib import Path
from datetime import datetime

import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter

# ---- Patterns (defined first on purpose) ----
DATE_RE = re.compile(r"^[A-Za-z]{3}\s+\d{1,2}$")
FILENAME_RE = re.compile(r"^(MV|Cash|FTP)_(\d{4})-(\d{2})-(\d{2})\.csv$", re.IGNORECASE)

# ---- Paths ----
PROJECT_DIR = Path(r"C:\Users\mkeita9\Projects\MOK_WBG_Projects\nav_project")
INPUT_DIR   = PROJECT_DIR / "input"
DIRS = {
    "MV":   INPUT_DIR / "mv",
    "Cash": INPUT_DIR / "cash",
    "FTP":  INPUT_DIR / "ftp",
}
REJECT_DIR  = INPUT_DIR / "rejected"
OUTPUT_DIR  = PROJECT_DIR / "output"

# ---- Aggregation level ----
GROUP_COL = "L4 Portfolio"

# ---- Which components are added up to get NAV ----
# Currently NAV = MV + Cash. Use ["MV", "Cash", "FTP"] if FTP is part of NAV.
NAV_COMPONENTS = ["MV", "Cash"]

# ---- Exclusion rules (column -> values to exclude; case and spaces ignored) ----
TYPOLOGY_EXCLUDE = ["FTP Loan Deposit"]

COUNTERPART_EXCLUDE = [          # MV ONLY
    "NW_LAM_REPO",
    "FTP_WASH",
    "LAM_STG_CB",
    "LAM_STG_ENHCASH",
    "LAM_STG_GRE",
    "LAM_STG_SP_INT",
    "LAM_STG_SP_USA",
    "LAM_STG_WTI",
    "POOL_ALM_COP",
    "NOSCNY",
    "FUNDXFER",
    "POOL_ALM_BRL",
    "POOL_LIQ_BRL",
    "POOL_LIQ_COP",
    "POOL_LIQ_INR",
    "POOL_LIQ_MXN",
    "POOL_LIQ_ZAR",
]

EXCLUDE_RULES = {
    "MV":   {"Typology": list(TYPOLOGY_EXCLUDE),
             "Counterpart": list(COUNTERPART_EXCLUDE)},
    "Cash": {"Typology": list(TYPOLOGY_EXCLUDE)},      # no Counterpart rule on Cash
    "FTP":  {},                                        # no filter on FTP
}
EXCLUDE_PREFIXES = {"MV": {}, "Cash": {}, "FTP": {}}

# Alternative column names per dataset (first match wins; case, spaces and symbols ignored)
COLUMN_ALIASES = {
    "MV":   {"Counterpart": ["Counterpart", "Counterparty"], "Typology": ["Typology"]},
    "Cash": {"Typology": ["Typology"]},
    "FTP":  {},
}

# ---- File reading ----
SEP = ","               # use ";" or "\t" if your CSVs need it
MOVE_REJECTED = False   # True = move wrongly named files to input/rejected

# ---- Output formatting ----
DECIMALS = 2
CHART_SCALE = 1         # 1_000_000 = chart axis in millions
CHART_UNIT = ""         # e.g. "(millions)"
pd.options.display.float_format = "{:,.2f}".format
pd.options.display.width = 200
pd.options.display.max_columns = 50

print("Project:", PROJECT_DIR)
for k, v in EXCLUDE_RULES.items():
    print(f"{k:5s} rules:", list(v) or "none")
print("NAV =", " + ".join(NAV_COMPONENTS))
```

```python
# ==== CELL 3: Helper functions ====
def clean_number(s: pd.Series) -> pd.Series:
    """'1,234.5' -> 1234.5 ; '(500)' -> -500 ; '-' or '' -> NaN"""
    s = s.astype(str).str.strip()
    neg = s.str.startswith("(") & s.str.endswith(")")
    s = s.str.replace(r"[(),\s]", "", regex=True)
    s = s.replace({"": None, "-": None, "nan": None, "None": None})
    out = pd.to_numeric(s, errors="coerce")
    return out.where(~neg, -out)

def norm_label(c) -> str:
    return re.sub(r"\s+", " ", str(c)).strip().rstrip(".")

def norm_text(s: pd.Series) -> pd.Series:
    """Text normalised for matching: no NBSP, trimmed, case-insensitive."""
    return (s.fillna("").astype(str).str.replace("\u00a0", " ", regex=False)
             .str.strip().str.casefold())

def _key(s):
    return re.sub(r"[^a-z0-9]", "", str(s).casefold())

def resolve_col(df, dataset, col):
    """Real column name for a rule column, using aliases."""
    names = [col] + list(COLUMN_ALIASES.get(dataset, {}).get(col, []))
    lookup = {_key(c): c for c in df.columns}
    for n in names:
        if _key(n) in lookup:
            return lookup[_key(n)]
    return None

CANONICAL = {"mv": "MV", "cash": "Cash", "ftp": "FTP"}

def dataset_of(path: Path) -> str:
    """'MV', 'Cash' or 'FTP', read from the file name."""
    m = FILENAME_RE.match(path.name)
    return CANONICAL[m.group(1).lower()]

def file_asof(path: Path):
    m = FILENAME_RE.match(path.name)
    return datetime(int(m.group(2)), int(m.group(3)), int(m.group(4))) if m else None

def valid_files(name: str, folder: Path):
    good = []
    for f in sorted(folder.glob("*.csv")):
        m = FILENAME_RE.match(f.name)
        if m and m.group(1).lower() == name.lower():
            good.append(f)
        else:
            print(f"REJECTED {f.name}: expected {name}_YYYY-MM-DD.csv")
            if MOVE_REJECTED:
                REJECT_DIR.mkdir(parents=True, exist_ok=True)
                f.rename(REJECT_DIR / f.name)
    return sorted(good, key=file_asof)

def parse_date_columns(cols, end_year):
    """Map 'Aug 21' headers to dates. Last column gets end_year;
    year steps back when months wrap past January."""
    parsed = [(c, datetime.strptime(norm_label(c), "%b %d"))
              for c in cols if DATE_RE.match(norm_label(c))]
    result, year = {}, end_year
    if not parsed:
        return result
    prev_month = parsed[-1][1].month
    for col, d in reversed(parsed):
        if d.month > prev_month:
            year -= 1
        prev_month = d.month
        result[col] = pd.Timestamp(year=year, month=d.month, day=d.day)
    return result

def exclusion_mask(df: pd.DataFrame, dataset: str, verbose=True, label="") -> pd.Series:
    """True for rows matching any exclusion rule of this dataset."""
    rules = EXCLUDE_RULES.get(dataset, {})
    prefixes = EXCLUDE_PREFIXES.get(dataset, {})
    total = pd.Series(False, index=df.index)
    for col in list(dict.fromkeys(list(rules) + list(prefixes))):
        real = resolve_col(df, dataset, col)
        if real is None:
            if verbose:
                print(f"     WARNING {label}: no column for '{col}' "
                      f"(columns: {list(df.columns)[:12]}...). Rule NOT applied")
            continue
        series = norm_text(df[real])
        m = pd.Series(False, index=df.index)
        values = {v.strip().casefold() for v in rules.get(col, [])}
        if values:
            m |= series.isin(values)
        for pre in prefixes.get(col, []):
            m |= series.str.startswith(pre.strip().casefold())
        if verbose:
            note = "  <-- 0 rows, check spelling" if int(m.sum()) == 0 else ""
            print(f"     {label}: {int(m.sum())} rows excluded by {col} (column '{real}'){note}")
        total |= m
    if verbose:
        if rules or prefixes:
            print(f"     {label}: {int(total.sum())} of {len(df)} rows excluded in total")
        else:
            print(f"     {label}: no exclusion rules, all {len(df)} rows kept")
    return total

def read_one_file(path: Path) -> pd.DataFrame:
    """One file -> long df (L4 Portfolio, Date, Value, source_file, file_time)."""
    dataset = dataset_of(path)
    df = pd.read_csv(path, sep=SEP, dtype=str, encoding_errors="replace")
    df.columns = [str(c).strip() for c in df.columns]
    if GROUP_COL not in df.columns:
        raise ValueError(f"'{GROUP_COL}' column not found. Columns: {list(df.columns)}")

    mask = exclusion_mask(df, dataset, label=f"[{dataset}] {path.name}")
    df = df[~mask].copy()

    asof = file_asof(path)
    date_cols = [c for c in df.columns if DATE_RE.match(norm_label(c))]
    if not date_cols:
        raise ValueError("no date columns detected")
    last_month = datetime.strptime(norm_label(date_cols[-1]), "%b %d").month
    end_year = asof.year - 1 if last_month > asof.month else asof.year
    date_map = parse_date_columns(list(df.columns), end_year)

    df[GROUP_COL] = df[GROUP_COL].fillna("UNKNOWN").str.strip()
    values = df[list(date_map)].apply(clean_number)
    values.columns = [date_map[c] for c in values.columns]
    values[GROUP_COL] = df[GROUP_COL]

    agg = values.groupby(GROUP_COL).sum(min_count=1)
    long = (agg.reset_index()
               .melt(id_vars=GROUP_COL, var_name="Date", value_name="Value")
               .dropna(subset=["Value"]))
    long["Date"] = pd.to_datetime(long["Date"])
    long["source_file"] = path.name
    long["file_time"] = asof
    return long

SIG_PATH = OUTPUT_DIR / "rules_signature.json"

def rules_signature(name: str) -> str:
    return json.dumps({"group": GROUP_COL,
                       "rules": EXCLUDE_RULES.get(name, {}),
                       "prefixes": EXCLUDE_PREFIXES.get(name, {}),
                       "aliases": COLUMN_ALIASES.get(name, {})}, sort_keys=True)

def check_history_rules(name: str, hist_path: Path):
    """If the rules changed since the history was saved, delete that history."""
    OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
    sigs = json.loads(SIG_PATH.read_text()) if SIG_PATH.exists() else {}
    if sigs.get(name) != rules_signature(name):
        if hist_path.exists():
            hist_path.unlink()
            print(f"[{name}] rules changed: old history deleted and rebuilt from the files")
        sigs[name] = rules_signature(name)
        SIG_PATH.write_text(json.dumps(sigs, indent=1))

def build_history(name: str, save=True) -> pd.DataFrame:
    """Read all valid files of a dataset, merge with saved history,
    latest as-of file wins per (L4 Portfolio, Date)."""
    folder = DIRS[name]
    files = valid_files(name, folder)
    if not files:
        raise FileNotFoundError(f"[{name}] no valid files in {folder}")
    hist_path = OUTPUT_DIR / f"{name.lower()}_history_long.csv"
    check_history_rules(name, hist_path)

    frames = []
    for f in files:
        try:
            part = read_one_file(f)
            frames.append(part)
            print(f"OK   [{name}] {f.name}: {part['Date'].nunique()} dates, "
                  f"{part[GROUP_COL].nunique()} {GROUP_COL} values")
        except Exception as e:
            print(f"SKIP [{name}] {f.name}: {e}")
    if not frames:
        raise RuntimeError(f"[{name}] no readable files")

    new = pd.concat(frames, ignore_index=True)
    if hist_path.exists():
        old = pd.read_csv(hist_path, parse_dates=["Date", "file_time"])
        if GROUP_COL not in old.columns:
            raise ValueError(f"{hist_path.name} was built with another grouping. "
                             "Run the reset cell, then rerun.")
        new = pd.concat([old, new], ignore_index=True)
    new = (new.sort_values("file_time", kind="stable")
              .drop_duplicates([GROUP_COL, "Date"], keep="last")
              .sort_values([GROUP_COL, "Date"]))
    if save:
        OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
        new.to_csv(hist_path, index=False)   # full precision kept on purpose
    return new

def to_wide(long: pd.DataFrame) -> pd.DataFrame:
    return long.pivot(index="Date", columns=GROUP_COL, values="Value").sort_index()

def latest_file(name: str) -> Path:
    files = valid_files(name, DIRS[name])
    print([f.name for f in files])
    if not files:
        raise FileNotFoundError(f"No {name}_YYYY-MM-DD.csv found in {DIRS[name]}")
    return files[-1]

print("Helpers loaded.")
```

```python
# ==== CELL 10: Compute NAV ====
WIDE = {"MV": mv, "Cash": cash, "FTP": ftp}

nav = pd.concat([WIDE[k].sum(axis=1, min_count=1).rename(k) for k in ["MV", "Cash", "FTP"]],
                axis=1).sort_index()
nav.index.name = "Date"

missing = nav[nav[NAV_COMPONENTS].isna().any(axis=1)]
if not missing.empty:
    print("WARNING: a NAV component is missing, NAV left blank for:")
    print(missing.index.strftime("%Y-%m-%d").tolist())

nav["NAV"] = nav[NAV_COMPONENTS].sum(axis=1, min_count=len(NAV_COMPONENTS))

# NAV by L4 Portfolio (a portfolio missing in one component counts as 0 there)
nav_by_l4 = reduce(lambda a, b: a.add(b, fill_value=0),
                   [WIDE[k] for k in NAV_COMPONENTS]).astype(float)
valid_dates = nav["NAV"].notna().reindex(nav_by_l4.index).fillna(False)
nav_by_l4 = nav_by_l4.where(valid_dates)

print("NAV =", " + ".join(NAV_COMPONENTS))
display(nav.tail(10))
```

```python
# ==== CELL 10b: NAV time series at L4 level = MV + Cash (formatted) ====
from openpyxl.styles import Font, PatternFill, Alignment, Border, Side
from openpyxl.utils import get_column_letter

# 1) NAV by L4 for every date present in both MV and Cash
common_dates = mv.index.intersection(cash.index)
nav_ts = mv.loc[common_dates].add(cash.loc[common_dates], fill_value=0)   # absent on one side counts as 0
nav_ts = nav_ts.astype(float).sort_index()
nav_ts.index.name = "Date"
nav_ts["TOTAL"] = nav_ts.sum(axis=1)

# 2) Check against total MV + total Cash
chk = (nav_ts["TOTAL"] - (mv.loc[common_dates].sum(axis=1) + cash.loc[common_dates].sum(axis=1))).abs().max()
print(f"Max reconciliation difference vs total MV + Cash: {chk:,.2f}")

# 3) Formatted display: all days (rows), L4 portfolios (columns)
display(nav_ts.style
        .format("{:,.2f}")
        .set_table_styles([{"selector": "th",
                            "props": [("background-color", "#1F3864"), ("color", "white"),
                                      ("text-align", "center")]}])
        .set_properties(subset=["TOTAL"], **{"font-weight": "bold"})
        .map(lambda v: "color: red" if v < 0 else ""))

# 4) Formatted Excel
NUM_FMT = '#,##0.00;[Red]-#,##0.00;"-"'
H_FILL = PatternFill("solid", start_color="1F3864")
H_FONT = Font(name="Calibri", bold=True, color="FFFFFF", size=11)
B_FONT = Font(name="Calibri", size=10)
BOLD   = Font(name="Calibri", size=10, bold=True)
BAND   = PatternFill("solid", start_color="F2F2F2")
LINE   = Border(bottom=Side(style="thin", color="BFBFBF"))

def style_ts(ws):
    for c in range(1, ws.max_column + 1):
        h = ws.cell(row=1, column=c)
        h.fill, h.font = H_FILL, H_FONT
        h.alignment = Alignment(horizontal="center", vertical="center", wrap_text=True)
    ws.row_dimensions[1].height = 32
    names = [ws.cell(row=1, column=c).value for c in range(1, ws.max_column + 1)]
    for r in range(2, ws.max_row + 1):
        for c in range(1, ws.max_column + 1):
            cell = ws.cell(row=r, column=c)
            cell.border = LINE
            if r % 2 == 1:
                cell.fill = BAND
            if c == 1:
                cell.number_format = "yyyy-mm-dd"
                cell.alignment = Alignment(horizontal="center")
                cell.font = B_FONT
            else:
                cell.number_format = NUM_FMT
                cell.alignment = Alignment(horizontal="right")
                cell.font = BOLD if names[c - 1] == "TOTAL" else B_FONT
    ws.column_dimensions["A"].width = 13
    for c in range(2, ws.max_column + 1):
        ws.column_dimensions[get_column_letter(c)].width = 22
    ws.freeze_panes = "B2"
    ws.auto_filter.ref = ws.dimensions
    ws.sheet_view.showGridLines = False

xlsx_l4 = OUTPUT_DIR / "nav_l4_timeseries.xlsx"
with pd.ExcelWriter(xlsx_l4, engine="openpyxl") as xw:
    nav_ts.round(DECIMALS).to_excel(xw, sheet_name="NAV_L4_TimeSeries", index_label="Date")
    mv.loc[common_dates].round(DECIMALS).to_excel(xw, sheet_name="MV_L4", index_label="Date")
    cash.loc[common_dates].round(DECIMALS).to_excel(xw, sheet_name="Cash_L4", index_label="Date")
    for s in ["NAV_L4_TimeSeries", "MV_L4", "Cash_L4"]:
        style_ts(xw.sheets[s])
print("Saved:", xlsx_l4)
```

```python
# ==== CELL 11: Checks ====
# 1) Reconciliation: sum of L4 portfolios must equal total NAV
chk = (nav_by_l4.sum(axis=1, min_count=1) - nav["NAV"]).abs().max()
print(f"Max reconciliation difference: {chk:,.2f}")

# 2) L4 Portfolios present in only some datasets
all_cols = {k: set(WIDE[k].columns) for k in WIDE}
union = set().union(*all_cols.values())
for k in WIDE:
    print(f"Missing in {k:5s}:", sorted(union - all_cols[k]))

# 3) Day-over-day change in %, to spot jumps or sign problems
display(nav.pct_change().tail(10).style.format("{:.2%}", na_rep="-"))
```

```python
# ==== CELL 12: Save outputs (formatted Excel + CSV, 2 decimals) ====
from openpyxl.styles import Font, PatternFill, Alignment, Border, Side
from openpyxl.utils import get_column_letter

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

mv.round(DECIMALS).to_csv(OUTPUT_DIR / "mv_timeseries_by_l4.csv", float_format="%.2f")
cash.round(DECIMALS).to_csv(OUTPUT_DIR / "cash_timeseries_by_l4.csv", float_format="%.2f")
ftp.round(DECIMALS).to_csv(OUTPUT_DIR / "ftp_timeseries_by_l4.csv", float_format="%.2f")
nav.round(DECIMALS).to_csv(OUTPUT_DIR / "nav_timeseries.csv", float_format="%.2f")

NUM_FMT  = '#,##0.00;[Red]-#,##0.00;"-"'
DATE_FMT = "yyyy-mm-dd"
HEADER_FILL = PatternFill("solid", start_color="1F3864")
HEADER_FONT = Font(name="Calibri", bold=True, color="FFFFFF", size=11)
BODY_FONT   = Font(name="Calibri", size=10)
BOLD_FONT   = Font(name="Calibri", size=10, bold=True)
THIN = Side(style="thin", color="BFBFBF")
BAND_FILL = PatternFill("solid", start_color="F2F2F2")

def format_sheet(ws, bold_cols=()):
    max_row, max_col = ws.max_row, ws.max_column
    for c in range(1, max_col + 1):
        cell = ws.cell(row=1, column=c)
        cell.fill, cell.font = HEADER_FILL, HEADER_FONT
        cell.alignment = Alignment(horizontal="center", vertical="center", wrap_text=True)
    ws.row_dimensions[1].height = 32
    header_names = [ws.cell(row=1, column=c).value for c in range(1, max_col + 1)]
    for r in range(2, max_row + 1):
        for c in range(1, max_col + 1):
            cell = ws.cell(row=r, column=c)
            cell.border = Border(bottom=THIN)
            if r % 2 == 1:
                cell.fill = BAND_FILL
            if c == 1:
                cell.number_format = DATE_FMT
                cell.alignment = Alignment(horizontal="center")
                cell.font = BODY_FONT
            else:
                cell.number_format = NUM_FMT
                cell.alignment = Alignment(horizontal="right")
                cell.font = BOLD_FONT if header_names[c - 1] in bold_cols else BODY_FONT
    ws.column_dimensions["A"].width = 13
    for c in range(2, max_col + 1):
        title_len = len(str(header_names[c - 1] or ""))
        ws.column_dimensions[get_column_letter(c)].width = max(18, min(title_len + 4, 40))
    ws.freeze_panes = "B2"
    ws.auto_filter.ref = ws.dimensions
    ws.sheet_view.showGridLines = False

xlsx_path = OUTPUT_DIR / "nav_report.xlsx"
with pd.ExcelWriter(xlsx_path, engine="openpyxl") as xw:
    nav.round(DECIMALS).to_excel(xw, sheet_name="NAV", index_label="Date")
    nav_by_l4.round(DECIMALS).to_excel(xw, sheet_name="NAV_by_L4", index_label="Date")
    mv.round(DECIMALS).to_excel(xw, sheet_name="MV_by_L4", index_label="Date")
    cash.round(DECIMALS).to_excel(xw, sheet_name="Cash_by_L4", index_label="Date")
    ftp.round(DECIMALS).to_excel(xw, sheet_name="FTP_by_L4", index_label="Date")
    format_sheet(xw.sheets["NAV"], bold_cols=("NAV",))
    for s in ["NAV_by_L4", "MV_by_L4", "Cash_by_L4", "FTP_by_L4"]:
        format_sheet(xw.sheets[s])
print("Saved to", OUTPUT_DIR)
```

```python
# ==== CELL 15: Diagnostic - reconcile one dataset with your Excel pivot ====
DATASET = "MV"            # "MV", "Cash" or "FTP"

# Excel Grand Totals (header label -> expected total). Empty dict = no comparison.
TARGETS = {
    "MV":   {"Aug 21": 639_740_452_023.10, "Oct 09": 627_058_673_952.81},
    "Cash": {},           # add Cash pivot totals, e.g. {"Aug 21": 123.45}
    "FTP":  {},
}
TARGET = TARGETS[DATASET]

path = latest_file(DATASET)
raw = pd.read_csv(path, sep=SEP, dtype=str, encoding_errors="replace")
raw.columns = [str(c).strip() for c in raw.columns]
print("File:", path.name, "| rows:", len(raw))

date_cols = [c for c in raw.columns if DATE_RE.match(norm_label(c))]
vals = raw[date_cols].apply(clean_number)
vals.columns = [norm_label(c) for c in date_cols]
last_lab = norm_label(date_cols[-1])

rows = []
labels = list(TARGET) if TARGET else [last_lab]
scenarios = {"No exclusion": pd.Series(False, index=raw.index)}
for col, vlist in EXCLUDE_RULES[DATASET].items():
    real = resolve_col(raw, DATASET, col)
    if real:
        scenarios[f"{col} only"] = norm_text(raw[real]).isin({v.strip().casefold() for v in vlist})
scenarios["All rules"] = exclusion_mask(raw, DATASET, verbose=False)
for name, m in scenarios.items():
    for lab in labels:
        tot = vals.loc[~m, lab].sum()
        row = {"Scenario": name, "Date": lab, "Total": tot, "Rows excluded": int(m.sum())}
        if lab in TARGET:
            row["Excel total"] = TARGET[lab]
            row["Difference"] = tot - TARGET[lab]
        rows.append(row)
display(pd.DataFrame(rows))
```

```python
# ==== CELL 17: Pull FX spot vs USD from FDR (Dremio REST API) ====
import os, time
import requests

BASE_URL = "https://findata-prod.worldbank.org:9047"      # PROD
TOKEN    = os.environ.get("FDR_TOKEN")
SCHEMA   = ["publishFINDATAREP", "IFCTRE", "Application", "Murex"]
if not TOKEN and not (OUTPUT_DIR / "fx_spot_raw.csv").exists():
    raise RuntimeError("Set the FDR_TOKEN environment variable first (see above).")
HEADERS = {"Authorization": f"Bearer {TOKEN}", "Content-Type": "application/json"}

LOCAL_PORTFOLIOS = ["POOL_ALM_BRL", "POOL_LIQ_BRL", "POOL_LIQ_COP",
                    "POOL_LIQ_INR", "POOL_LIQ_MXN", "POOL_LIQ_ZAR"]
# "LOCAL_PER_USD": USD/BRL = 5.40 -> NAV_USD = NAV_local / rate
# "USD_PER_LOCAL": BRL/USD = 0.185 -> NAV_USD = NAV_local * rate
FX_QUOTE = "LOCAL_PER_USD"
USE_CACHE = False        # True = reuse output/fx_spot_raw.csv, no connection needed

def ccy_of(portfolio: str) -> str:
    return portfolio.strip()[-3:].upper()      # POOL_LIQ_BRL -> BRL

NEEDED_CCY = sorted({ccy_of(p) for p in LOCAL_PORTFOLIOS})
start = (nav_ts.index.min() - pd.Timedelta(days=10)).strftime("%Y-%m-%d")  # buffer for holidays
end   = nav_ts.index.max().strftime("%Y-%m-%d")

SQL_QUERY = (
    'SELECT "Business_Date", "MX_BASE_CURR", "MX_LABEL", "MX_MID", "MX_UNDERLYIN" '
    'FROM "Fx_Spot_Detail" '
    "WHERE \"MX_BASE_CURR\" = 'USD' "
    f"AND \"Business_Date\" >= CAST('{start}' AS DATE) "
    f"AND \"Business_Date\" <= CAST('{end}' AS DATE)"
)

def run_dremio_query(sql, page=500, timeout_s=300):
    r = requests.post(f"{BASE_URL}/api/v3/sql", headers=HEADERS,
                      json={"sql": sql, "context": SCHEMA}, timeout=60)
    r.raise_for_status()
    job_id = r.json()["id"]
    t0 = time.time()
    while True:
        j = requests.get(f"{BASE_URL}/api/v3/job/{job_id}", headers=HEADERS, timeout=60).json()
        state = j.get("jobState")
        if state == "COMPLETED":
            break
        if state in ("FAILED", "CANCELED", "CANCELLED"):
            raise RuntimeError(f"Query {state}: {j.get('errorMessage', j)}")
        if time.time() - t0 > timeout_s:
            raise TimeoutError("Query timed out")
        time.sleep(1)
    total = j.get("rowCount", 0)
    rows, offset = [], 0
    while offset < total:
        res = requests.get(f"{BASE_URL}/api/v3/job/{job_id}/results",
                           headers=HEADERS, params={"offset": offset, "limit": page}, timeout=60)
        res.raise_for_status()
        chunk = res.json().get("rows", [])
        if not chunk:
            break
        rows.extend(chunk)
        offset += len(chunk)
    return pd.DataFrame(rows)

cache = OUTPUT_DIR / "fx_spot_raw.csv"
if USE_CACHE and cache.exists():
    fx_raw = pd.read_csv(cache, parse_dates=["Business_Date"])
else:
    fx_raw = run_dremio_query(SQL_QUERY)
    fx_raw.to_csv(cache, index=False)

print("Rows pulled:", len(fx_raw))
fx_raw["Business_Date"] = pd.to_datetime(fx_raw["Business_Date"])
fx_raw["ccy"] = fx_raw["MX_UNDERLYIN"].astype(str).str.strip().str.upper()
fx_raw["MX_MID"] = pd.to_numeric(fx_raw["MX_MID"], errors="coerce")
print("Distinct MX_UNDERLYIN:", sorted(fx_raw["ccy"].unique())[:40])

sel = fx_raw[fx_raw["ccy"].isin(NEEDED_CCY)]
dups = sel.duplicated(["Business_Date", "ccy"]).sum()
if dups:
    print(f"WARNING: {dups} duplicate (date, currency) rows. The last one is kept. Check MX_LABEL.")
fx = (sel.sort_values("Business_Date")
         .drop_duplicates(["Business_Date", "ccy"], keep="last")
         .pivot(index="Business_Date", columns="ccy", values="MX_MID").sort_index())
fx.index.name = "Date"
print("Missing currencies:", sorted(set(NEEDED_CCY) - set(fx.columns)))
display(fx.tail(5).style.format("{:,.4f}"))
```

```python
# ==== CELL 18: NAV in USD at L4 level ====
# Align FX on NAV dates; weekends/holidays take the last available rate
fx_all = fx.reindex(fx.index.union(nav_ts.index)).sort_index()
fx_aligned = fx_all.ffill().reindex(nav_ts.index)
filled = fx_all.reindex(nav_ts.index).isna().sum()
print("Dates where the previous FX rate was carried forward:\n", filled.to_string())

nav_local = nav_ts.drop(columns="TOTAL")
nav_usd = nav_local.copy()
fx_by_portfolio = pd.DataFrame(index=nav_ts.index)

for p in LOCAL_PORTFOLIOS:
    col = next((c for c in nav_local.columns if str(c).strip().casefold() == p.casefold()), None)
    if col is None:
        print(f"NOTE: {p} is not in the NAV table, skipped")
        continue
    ccy = ccy_of(p)
    if ccy not in fx_aligned.columns:
        print(f"WARNING: no FX for {ccy}, {p} left unconverted")
        continue
    rate = fx_aligned[ccy]
    fx_by_portfolio[col] = rate
    nav_usd[col] = nav_local[col] / rate if FX_QUOTE == "LOCAL_PER_USD" else nav_local[col] * rate

nav_usd["TOTAL_USD"] = nav_usd.sum(axis=1)
nav_usd.index.name = "Date"

# Sanity check: rates should look plausible (e.g. USD/BRL about 5, USD/COP about 4,000)
print("Latest rate used:\n", fx_aligned.iloc[-1].round(4).to_string())
display(nav_usd.tail(10).style.format("{:,.2f}"))
```

```python
# ==== CELL 19: Create the USD report file (NAV L4 + FX_Spot + NAV_USD tabs) ====
import shutil

SRC_XLSX = OUTPUT_DIR / "nav_l4_timeseries.xlsx"            # written by Cell 10b
NEW_XLSX = OUTPUT_DIR / "NAV_L4_Local_and_USD_Report.xlsx"  # new file name
if not SRC_XLSX.exists():
    raise FileNotFoundError("Run Cell 10b first so the source file exists.")
shutil.copyfile(SRC_XLSX, NEW_XLSX)

with pd.ExcelWriter(NEW_XLSX, engine="openpyxl", mode="a", if_sheet_exists="replace") as xw:
    fx_aligned.to_excel(xw, sheet_name="FX_Spot", index_label="Date")
    nav_usd.round(DECIMALS).to_excel(xw, sheet_name="NAV_USD", index_label="Date")
    style_ts(xw.sheets["FX_Spot"])
    style_ts(xw.sheets["NAV_USD"])
    for row in xw.sheets["FX_Spot"].iter_rows(min_row=2, min_col=2):
        for cell in row:
            cell.number_format = "#,##0.0000"        # FX rates need 4 decimals
    for cell in xw.sheets["NAV_USD"][1]:
        if cell.value == "TOTAL_USD":
            cell.value = "TOTAL USD"
print("Created:", NEW_XLSX)
```
