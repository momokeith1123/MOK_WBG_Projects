```python
import win32com.client

outlook = win32com.client.GetActiveObject("Outlook.Application")
namespace = outlook.GetNamespace("MAPI")
namespace.Logon()
inbox = namespace.GetDefaultFolder(6)
ice_folder = inbox.Folders["14.ICE"]

messages = ice_folder.Items
messages.Sort("[ReceivedTime]", True)

print(f"Total emails in 14.ICE: {messages.Count}\n")

for i, msg in enumerate(messages):
    if i >= 10:  # Print first 10 only
        break
    smtp  = getattr(msg, "SenderEmailAddress", "N/A")
    name  = getattr(msg, "SenderName", "N/A")
    subj  = getattr(msg, "Subject", "N/A")
    print(f"[{i+1}] From: {smtp} | Name: {name} | Subject: {subj}")
```
